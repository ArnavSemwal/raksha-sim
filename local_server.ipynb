```python
from flask import Flask, request, jsonify
from triage_engine import analyze_patient

app = Flask(__name__)

@app.route('/predict', methods=['POST'])
def predict():
    sensor_packet = request.json
    print(f"\n[RECEIVED] {sensor_packet.get('patient_id', 'Unknown')} Data")
    
    try:
        ai_result = analyze_patient(sensor_packet)
        print(f"[AI DECISION] {ai_result}")
        return jsonify(ai_result), 200
        
    except Exception as e:
        print(f"[ERROR] {e}")
        return jsonify({"error": str(e)}), 500

if __name__ == '__main__':
    print("Starting AI Local Server...")
    print("Listening for Shashwat's simulator on: http://127.0.0.1:5000/predict")
    app.run(host='0.0.0.0', port=5000)
```
